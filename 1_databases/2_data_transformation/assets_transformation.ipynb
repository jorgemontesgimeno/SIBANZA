```python
import pandas as pd

assets_df = pd.read_csv("../1_data_extraction/data/assets_info.csv")

assets_df["score_PER"] = (1/assets_df["trailingpe"] - (1/assets_df["trailingpe"]).mean()) / (1/assets_df["trailingpe"]).std()
assets_df["score_PB"] = (1/assets_df["pricetobook"] - (1/assets_df["pricetobook"]).mean()) / (1/assets_df["pricetobook"]).std()
assets_df["score_div"] = (assets_df["dividend_yield"] - assets_df["dividend_yield"].mean()) / assets_df["dividend_yield"].std()
assets_df["score_fcf"] = (assets_df["fcf_yield"] - assets_df["fcf_yield"].mean()) / assets_df["fcf_yield"].std()

assets_df["value_score"] = (
    assets_df["score_PER"] +
    assets_df["score_PB"] +
    assets_df["score_div"] +
    assets_df["score_fcf"]
)

## Las columnas con mayor value_score son las empresas mas infravaloradas

assets_df["rev_z"] = (assets_df["revenuegrowth"] - assets_df["revenuegrowth"].mean()) / assets_df["revenuegrowth"].std()
assets_df["earn_z"] = (assets_df["earningsgrowth"] - assets_df["earningsgrowth"].mean()) / assets_df["earningsgrowth"].std()

assets_df["growth_score"] = (
    assets_df["rev_z"] +
    assets_df["earn_z"]
)

# Las columnas con mayor growth_score son las empresas con mayor crecimiento

assets_df["dividend_score"] = (
assets_df["dividend_yield"] *
(1 - assets_df["payoutratio"]) *
assets_df["fcf_yield"]
)

# Las columnas con mayor dividend_score son las empresas con mayor potencial de dividendos

assets_df = assets_df.drop(columns=["score_PER", "score_PB", "score_div", "score_fcf"])

assets_df = assets_df.drop(columns=["rev_z", "earn_z"])

assets_df.to_csv("../1_data_extraction/data/assets_info.csv", index=False)
```
