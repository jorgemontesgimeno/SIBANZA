```python
import pandas as pd

df = pd.read_csv("../1_data_extraction/data/historical_assets.csv")

df["date"] = pd.to_datetime(df["date"])

df["return_3m"] = df.groupby("ticker")["Close"].pct_change(63)
df["return_6m"] = df.groupby("ticker")["Close"].pct_change(126)

df["momentum_score"] = df["return_3m"] + df["return_6m"]

df["momentum_score"] = (
    0.5 * df["return_3m"] +
    0.5 * df["return_6m"]
)

df["returns"] = df.groupby("ticker")["Close"].pct_change()
df["volatility"] = df.groupby("ticker")["returns"].rolling(63).std().reset_index(level=0, drop=True)

df["low_vol_score"] = -df["volatility"]



df.to_csv("../1_data_extraction/data/historical_assets.csv", index=False)


```
