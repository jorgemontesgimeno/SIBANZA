```python
import pandas as pd
import numpy as np
from scipy.optimize import minimize

# ============================================================
# 1. CARGAR DATOS
# ============================================================

assets_info = pd.read_csv(
    "results/final_scores.csv"
)

historical_assets = pd.read_csv(
    "../../1_databases/1_data_extraction/data/historical_assets.csv"
)

ml_predictions = pd.read_csv(
    "results/random_forest_selected_assets_threshold.csv"
)

selected_tickers = ml_predictions["asset"].unique()

scores_df = assets_info[
    assets_info["ticker"].isin(selected_tickers)
]

ml_predictions.to_csv("df_preparados/ml_predictions.csv", index=False)
scores_df.to_csv("df_preparados/scores_df.csv", index=False)


historical_assets = pd.read_csv(
    "../../1_databases/1_data_extraction/data/historical_assets.csv"
)

# Crear matriz de precios
prices = historical_assets.pivot(
    index="date",
    columns="ticker",
    values="Close"
)

# Ordenar fechas
prices = prices.sort_index()

# Crear retornos
returns = prices.pct_change()

# Eliminar NaN
returns = returns.dropna(how="all")

returns = returns[ml_predictions["asset"]]

returns.to_csv("df_preparados/returns.csv", index=True)
```
