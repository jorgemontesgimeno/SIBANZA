```python
import pandas as pd

# =========================
# CARGA DE DATOS
# =========================

assets_info = pd.read_csv(
    "../../1_databases/1_data_extraction/data/assets_info.csv"
)

historical_assets = pd.read_csv(
    "../../1_databases/1_data_extraction/data/historical_assets.csv"
)

# =========================
# SCORES DE assets_info
# =========================

scores_info = assets_info[
    [
        "ticker",
        "dividend_score",
        "growth_score",
        "value_score"
    ]
]

# =========================
# LOW VOL SCORE
# =========================

low_vol = historical_assets[
    ["date", "ticker", "low_vol_score"]
]

# Nos quedamos con el último score disponible por ticker
low_vol = (
    low_vol
    .sort_values("date")
    .groupby("ticker")
    .tail(1)
)

# =========================
# MOMENTUM SCORE
# =========================

momentum = historical_assets[
    ["date", "ticker", "momentum_score"]
]

# Último momentum disponible
momentum = (
    momentum
    .sort_values("date")
    .groupby("ticker")
    .tail(1)
)

# =========================
# MERGE FINAL
# =========================

final_scores = (
    scores_info
    .merge(
        low_vol[["ticker", "low_vol_score"]],
        on="ticker",
        how="left"
    )
    .merge(
        momentum[["ticker", "momentum_score"]],
        on="ticker",
        how="left"
    )
)

# =========================
# RESULTADO
# =========================

final_scores.to_csv("results/final_scores.csv", index=False)
```
