```python
import pandas as pd

df = pd.read_csv("../../1_databases/1_data_extraction/data/historical_assets.csv")

low_vol_score = (
    df.sort_values(["date", "volatility"], ascending=[True, True])
      .groupby("date")
      .head(10)
)

low_vol_score = low_vol_score[["date", "ticker", "low_vol_score"]]

low_vol_score.dropna(inplace=True)

print("LOW VOLATILITY: ")
print("")
print("Fecha minima:", low_vol_score["date"].min())
print("Fecha maxima:", low_vol_score["date"].max())
print("")

print("Top 5 del 16/12/2022:")
print(low_vol_score[low_vol_score["date"] == "2022-12-16"].head(5))
```
