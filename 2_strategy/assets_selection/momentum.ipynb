```python
import pandas as pd

df = pd.read_csv("../../1_databases/1_data_extraction/data/historical_assets.csv")


top_momentum = (
    df.sort_values(["date", "momentum_score"], ascending=[True, False])
      .groupby("date")
      .head(10)
)

top_momentum = top_momentum[["date", "ticker", "momentum_score"]]

top_momentum.dropna(inplace=True)

print("MOMENTUM: ")
print("")
print("Fecha minima:", top_momentum["date"].min())
print("Fecha maxima:", top_momentum["date"].max())
print("")

print("Top 5 del 16/12/2022:")
print(top_momentum[top_momentum["date"] == "2022-12-16"].head(5))
```
